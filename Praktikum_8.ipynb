## 1. Dasar Teori

### 1.1 List
List merupakan struktur data di Python yang digunakan untuk menyimpan sekumpulan data. Karakteristik utamanya adalah **mutable** (bisa diubah setelah dibuat), sehingga sangat fleksibel untuk operasi tambah, hapus, maupun ubah elemen.

### 1.2 Tuple
Tuple mirip dengan list, namun bersifat **immutable** (tidak bisa diubah setelah dibuat). Tuple cocok digunakan untuk data yang sifatnya tetap, seperti kode barang atau ID produk.

### 1.3 Dictionary
Dictionary menyimpan data dalam pasangan **key-value**. Struktur ini sangat berguna untuk menghubungkan nama barang dengan jumlah stoknya, harga, dan atribut lainnya.

## 2. Praktikum Struktur Data

### 2.1 Praktikum List

Saya mulai dengan membuat list sederhana berisi daftar barang gudang, kemudian menambahkan item baru menggunakan method `.append()`.

```python
# Membuat list barang
barang = ["Laptop", "Mouse", "Keyboard"]

# Menampilkan data awal
print("Daftar Barang:")
for item in barang:
    print(f"  - {item}")

# Menambahkan barang baru
barang.append("Printer")

print("\nSetelah ditambah 'Printer':")
print(barang)
```

**Penjelasan:**  
Dari output di atas, terlihat bahwa method `.append()` berhasil menambahkan elemen `"Printer"` ke akhir list. Ini menunjukkan sifat **mutable** dari list — saya bisa menambah data kapan saja tanpa membuat ulang variabelnya.

### 2.2 Praktikum Tuple

Tuple saya gunakan untuk menyimpan kode barang yang sifatnya tidak boleh berubah selama program berjalan.

```python
# Tuple kode barang
kode_barang = ("BRG001", "BRG002", "BRG003")

print("Kode Barang:")
for kode in kode_barang:
    print(f"  [{kode}]")

# Membuktikan bahwa tuple immutable
try:
    kode_barang[0] = "BRG999"
except TypeError as e:
    print(f"\nError saat mengubah tuple: {e}")
    print("Ini membuktikan bahwa tuple bersifat immutable.")
```

**Penjelasan:**  
Saya sengaja mencoba mengubah elemen tuple untuk membuktikan bahwa tuple memang **immutable**. Hasilnya, Python mengeluarkan `TypeError` yang mengonfirmasi bahwa data dalam tuple tidak bisa dimodifikasi. Dalam konteks industri, ini penting untuk menjaga integritas data seperti kode barang atau nomor seri.

### 2.3 Praktikum Dictionary

Dictionary sangat cocok untuk data stok barang karena setiap barang (key) memiliki jumlah stok (value) yang berpasangan.

```python
# Data stok barang menggunakan dictionary
stok_barang = {
    "Laptop": 10,
    "Mouse": 25,
    "Keyboard": 15
}

# Menampilkan stok
print("=" * 35)
print("     DATA STOK BARANG GUDANG")
print("=" * 35)
print(f"{'Nama Barang':<15} {'Stok':>10}")
print("-" * 35)
for barang, stok in stok_barang.items():
    print(f"{barang:<15} {stok:>10} unit")
print("-" * 35)
print(f"{'Total item':<15} {len(stok_barang):>10} jenis")
```

## 3. Studi Kasus: Sistem Pengolahan Data Stok Barang

Pada bagian ini, saya membuat program pengolahan stok barang yang lebih lengkap — mencakup fitur input data, tampil data, hapus barang, update stok, serta perhitungan harga dan total nilai stok.

> **Catatan:** Karena notebook tidak mendukung `input()` secara interaktif dalam semua environment, saya menggunakan data yang sudah didefinisikan langsung.

```python
# Sistem Pengolahan Data Stok Barang (Lengkap)

stok_barang = {}

# Simulasi input data (biasanya pakai input(), tapi di notebook didefinisikan langsung)
data_input = [
    {"nama": "Laptop", "stok": 10, "harga": 7000000},
    {"nama": "Mouse", "stok": 20, "harga": 150000},
    {"nama": "Printer", "stok": 5, "harga": 2500000},
    {"nama": "Keyboard", "stok": 15, "harga": 350000},
]

# Memasukkan data ke dictionary
for item in data_input:
    stok_barang[item["nama"]] = {"stok": item["stok"], "harga": item["harga"]}

# Menampilkan hasil
print("=" * 60)
print("           LAPORAN DATA STOK BARANG GUDANG")
print("=" * 60)
print(f"{'Barang':<12} {'Stok':>6} {'Harga (Rp)':>15} {'Total Nilai':>18}")
print("-" * 60)

total_nilai_keseluruhan = 0
for barang, info in stok_barang.items():
    total_nilai = info["stok"] * info["harga"]
    total_nilai_keseluruhan += total_nilai
    print(f"{barang:<12} {info['stok']:>6} {info['harga']:>15,} {total_nilai:>18,}")

print("-" * 60)
print(f"{'TOTAL':>35} {total_nilai_keseluruhan:>18,}")
print("=" * 60)
```

### 3.1 Fitur Hapus dan Update Stok Barang

Selanjutnya saya menambahkan fitur untuk menghapus dan mengupdate data stok.

```python
# Fitur Update Stok
print("--- UPDATE STOK ---")
barang_update = "Mouse"
stok_baru = 30
stok_barang[barang_update]["stok"] = stok_baru
print(f"Stok '{barang_update}' berhasil diupdate menjadi {stok_baru} unit\n")

# Fitur Hapus Barang
print("--- HAPUS BARANG ---")
barang_hapus = "Printer"
if barang_hapus in stok_barang:
    del stok_barang[barang_hapus]
    print(f"Barang '{barang_hapus}' berhasil dihapus\n")

# Tampilkan data setelah update & hapus
print("=" * 60)
print("     DATA STOK SETELAH UPDATE & HAPUS")
print("=" * 60)
print(f"{'Barang':<12} {'Stok':>6} {'Harga (Rp)':>15} {'Total Nilai':>18}")
print("-" * 60)

total_nilai_keseluruhan = 0
for barang, info in stok_barang.items():
    total_nilai = info["stok"] * info["harga"]
    total_nilai_keseluruhan += total_nilai
    print(f"{barang:<12} {info['stok']:>6} {info['harga']:>15,} {total_nilai:>18,}")

print("-" * 60)
print(f"{'TOTAL':>35} {total_nilai_keseluruhan:>18,}")
```

## 4. Pengolahan File CSV Menggunakan Pandas

Pada bagian ini saya akan membuat file CSV produksi, kemudian membacanya menggunakan Pandas, dan melakukan analisis data sederhana.

### 4.1 Membuat dan Membaca File CSV

```python
import pandas as pd

# Membuat file CSV terlebih dahulu
data_csv = """Tanggal,Produk,Jumlah
2026-01-01,A,100
2026-01-02,B,120
2026-01-03,C,90
2026-01-04,D,150
2026-01-05,E,110"""

with open("produksi.csv", "w") as f:
    f.write(data_csv)

print("File 'produksi.csv' berhasil dibuat!\n")

# Membaca file CSV
data = pd.read_csv("produksi.csv")

# Menampilkan data
print("Data Produksi:")
print(data)
```

### 4.2 Menampilkan Informasi Data

Sebelum melakukan analisis, penting untuk memahami struktur data yang saya miliki — termasuk jumlah baris, kolom, tipe data, dan statistik deskriptifnya.

```python
# Jumlah Baris dan Kolom
print("Jumlah Baris dan Kolom:", data.shape)
print()

# Informasi Data
print("Informasi Data:")
print(data.info())
print()

# Statistik Deskriptif
print("Statistik Deskriptif:")
print(data.describe())
```

### 4.3 Filter Data Produksi

Saya melakukan filtering untuk menampilkan produk dengan jumlah produksi di atas 100 unit.

```python
# Filter data produksi di atas 100
hasil = data[data["Jumlah"] > 100]

print("Produk dengan Produksi > 100 unit:")
print(hasil)
```

### 4.4 Menambahkan Data Baru dan Menyimpan ke CSV

```python
# Menambahkan data baru
data_baru = pd.DataFrame({
    "Tanggal": ["2026-01-06"],
    "Produk": ["F"],
    "Jumlah": [135]
})

data = pd.concat([data, data_baru], ignore_index=True)
print("Data setelah ditambah:")
print(data)

# Simpan ke file CSV baru
data.to_csv("hasil_produksi.csv", index=False)
print("\nFile 'hasil_produksi.csv' berhasil disimpan!")
```

## 5. Analisis Data Produksi

Sekarang saya melakukan analisis statistik sederhana terhadap data produksi — meliputi total, rata-rata, produksi terbesar, dan terkecil.

```python
# Analisis Data Produksi
total = data["Jumlah"].sum()
rata_rata = data["Jumlah"].mean()
maksimum = data["Jumlah"].max()
minimum = data["Jumlah"].min()
produk_tertinggi = data.loc[data["Jumlah"].idxmax(), "Produk"]
produk_terendah = data.loc[data["Jumlah"].idxmin(), "Produk"]

print("=" * 40)
print("     ANALISIS DATA PRODUKSI")
print("=" * 40)
print(f"Total Produksi      : {total} unit")
print(f"Rata-rata Produksi  : {rata_rata:.1f} unit")
print(f"Produksi Terbesar   : {maksimum} unit (Produk {produk_tertinggi})")
print(f"Produksi Terkecil   : {minimum} unit (Produk {produk_terendah})")
print("=" * 40)
```

## 6. Visualisasi Data Produksi

Untuk mempermudah pemahaman data, saya membuat grafik batang yang menampilkan jumlah produksi tiap produk.

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8, 5))

colors = ['#2196F3', '#4CAF50', '#FF9800', '#E91E63', '#9C27B0', '#00BCD4']
bars = ax.bar(data["Produk"], data["Jumlah"], color=colors, edgecolor='white', linewidth=1.2)

# Menambahkan label nilai di atas bar
for bar, val in zip(bars, data["Jumlah"]):
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 2,
            str(val), ha='center', va='bottom', fontweight='bold', fontsize=11)

ax.set_title("Grafik Produksi per Produk", fontsize=14, fontweight='bold', pad=15)
ax.set_xlabel("Produk", fontsize=12)
ax.set_ylabel("Jumlah Produksi (unit)", fontsize=12)
ax.set_ylim(0, max(data["Jumlah"]) * 1.15)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```
